## Colab sürümü

Colab çıktısı imleci istediğimiz yere götüremediği için burada ekranı her karede baştan çiziyoruz. Ekranı satırlardan oluşan bir liste, her satırı da karakterlerden oluşan bir liste olarak tutuyoruz (iç içe liste).

```python
import random
import time
from IPython.display import clear_output

KARAKTERLER = "ｱｲｳｴｵｶｷｸｹｺｻｼｽｾｿﾀﾁﾂﾃﾄﾅﾆﾇﾈﾉﾊﾋﾌﾍﾎﾏﾐﾑﾒﾓ0123456789"
YESIL = "\033[32m"
SIFIRLA = "\033[0m"


def colab_yagmur(genislik=50, yukseklik=20, kare=150, hiz=0.08, kuyruk=10):
    damlalar = [random.randint(-yukseklik, 0) for _ in range(genislik)]
    izgara = [[" "] * genislik for _ in range(yukseklik)]

    for _ in range(kare):
        for x in range(genislik):
            y = damlalar[x]
            if 0 <= y < yukseklik:
                izgara[y][x] = random.choice(KARAKTERLER)   # damlanın başı
            if 0 <= y - kuyruk < yukseklik:
                izgara[y - kuyruk][x] = " "                 # kuyruğun ucu silinir
            damlalar[x] += 1
            if y - kuyruk >= yukseklik:                      # damla ekrandan çıktı
                damlalar[x] = random.randint(-yukseklik, 0)

        clear_output(wait=True)
        print(YESIL + "\n".join("".join(satir) for satir in izgara) + SIFIRLA)
        time.sleep(hiz)
```

```python
colab_yagmur()
```

## Terminal sürümü

Bu kod Colab'da değil, kendi bilgisayarının terminalinde çalışır. Hücreyi çalıştırma; kodu kopyalayıp `matrix.py` adıyla kaydet ve terminalde `python matrix.py` yaz. Mac/Linux terminali ve Windows Terminal'de çalışır.

```python
import os
import random
import sys
import time

KARAKTERLER = "ｱｲｳｴｵｶｷｸｹｺｻｼｽｾｿﾀﾁﾂﾃﾄﾅﾆﾇﾈﾉﾊﾋﾌﾍﾎﾏﾐﾑﾒﾓ0123456789"

YESIL = "\033[32m"           # yazı rengi: yeşil
BEYAZ = "\033[97m"           # yazı rengi: parlak beyaz
SIFIRLA = "\033[0m"          # renkleri sıfırla
TEMIZLE = "\033[2J"          # ekranı temizle
IMLEC_GIZLE = "\033[?25l"
IMLEC_GOSTER = "\033[?25h"


def yaz(satir, sutun, metin):
    """İmleci (satir, sutun) konumuna götürüp metni oraya yazar."""
    sys.stdout.write(f"\033[{satir};{sutun}H{metin}")


def yagmur(hiz=0.05, kuyruk=12):
    genislik, yukseklik = os.get_terminal_size()
    yukseklik -= 1                                   # en alt satıra yazıp ekranı kaydırmayalım
    damlalar = [random.randint(-yukseklik, 0) for _ in range(genislik)]

    sys.stdout.write(TEMIZLE + IMLEC_GIZLE)
    try:
        while True:
            for x in range(genislik):
                y = damlalar[x]                      # bu sütundaki damlanın başı
                if 1 <= y <= yukseklik:
                    yaz(y, x + 1, BEYAZ + random.choice(KARAKTERLER))
                if 1 <= y - 1 <= yukseklik:
                    yaz(y - 1, x + 1, YESIL + random.choice(KARAKTERLER))
                if 1 <= y - kuyruk <= yukseklik:
                    yaz(y - kuyruk, x + 1, " ")      # kuyruğun ucunu sil

                damlalar[x] += 1
                if y - kuyruk > yukseklik:           # damla ekranın altından çıktı
                    damlalar[x] = random.randint(-yukseklik, 0)
            sys.stdout.flush()
            time.sleep(hiz)
    except KeyboardInterrupt:                        # Ctrl+C ile çıkış
        pass
    finally:
        sys.stdout.write(SIFIRLA + IMLEC_GOSTER + TEMIZLE + "\033[H")


if __name__ == "__main__":
    yagmur()
```
